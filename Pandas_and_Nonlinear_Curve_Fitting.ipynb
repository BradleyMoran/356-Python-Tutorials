# Non-linear curve-fitting 

## Pandas Dataframes
Pandas is an important Python library for data science.  If you are at all interested in this topic, you should learn more about Pandas (short for Python Data Analysis). Here, we will just scratch the surface, using Pandas to examine our data and port into our script. The data structure we use is called dataframe, or df for short. It is similar to a spreadsheet. Note that there is a column of integers on the left margin called the index.  We can refer to a column using the name in the column head. For instance, in the dataframe below ``df['t(s)]`` would give you all the values in the time column.

```python
import pandas as pd
df=pd.read_csv('files/Shock absorber.csv')
print(df)
```

## Plotting
We can plot directly from the dataframe.

```python
import matplotlib.pyplot as plt
plt.plot(df['t (s)'], df['y (mm)'],'o')  # note the syntax here:
plt.ylabel('y (mm)')
plt.xlabel('t(s)')
```

## Curve Fitting
The scatterplot above suggests that our data could be fit to a dampened occillation equation:  

$y=Ae^{-Bx}sin(Cx)$

Here A, B, and C are constants.


```python
import numpy as np                        # we need numpy for the math functions
from scipy.optimize import curve_fit     # we need scipy for the fitting functions



# Function to model and fit: Put your function here
def my_function(t, A, B, C):    # we list the independent variables and the fit parameters here
    return A * np.exp(-B * t)*np.sin(C*t) # note that sin and exp don't exist in Python; we have to get them from numpy


# Selecting data from the dataframe

# the .to_numpy() command changes the series from the dataframe into a numpy array
t_data = df['t (s)'].to_numpy()   # we could just call this x, but we should use the actual variable name
y_data = df['y (mm)'].to_numpy()  # we could just call this y, but we should use the actual variable name


# Add a guess value for the fit parameters you intend to fit. (Not always necessary but can speed up fit and assure correct values are determined)
guess=(100, 0.5, 1)       #your guess value for A, B,  and C. These should be specified in the order in which they were defined earlier.


# Fit the model to the data
parameters, covariance = curve_fit(my_function, t_data, y_data, p0=guess)

# parameters contains the best-fit parameters for A, B, and C
A, B, C = parameters

# covariance is the covariance matrix; the square root of its diagonal elements gives the uncertainties
A_error, B_error, C_error = np.sqrt(np.diag(covariance))

# Display the results
print("Fitted parameters: A =", A, "   B =", B, "   C =", C)
print("Uncertainties: error in A =", A_error, "   error in B =", B_error, "   error in C =", C_error)
```

## Rounding Data
Note rounding and units:  

$A = 100.000\pm 0.040$ (mm)  

$B=0.19994 \pm 0.00011$ (1/s)  

$C=1.000000 \pm 0.000062$ (1/s)



## Plotting Fit
It is always a good idea to add your fit to the scatterplot to get a visual confirmation of the fit.  Use a line for the fit and markers for experimental data.

```python
plt.plot(t_data, my_function(t_data,A,B,C))
```

## Make it Nice
This is not very nice-the points are too far apart.  For your fit you should generate more closely-spaced x values.

Below, we generate *many* x values and then use the parameters from our plot to make a smooth curve.

```python
t_fit=np.linspace(1,13,100) # Here we make 100 points between 1 and 13
plt.plot(t_fit,my_function(t_fit,A,B,C))
```

## Putting it All Together
We should put it all on one plot and add labels and a legend.

```python
plt.plot(df['t (s)'], df['y (mm)'],'o',label='Experimental')  # experimental data

# plot of fit
t_fit=np.arange(1,13,0.1)    # generates a list from 1 to just short of 13 in steps of 0.1
plt.plot(t_fit, my_function(t_fit,A,B,C),label='Fit')

# add labels and a legend
plt.ylabel('y (mm)')
plt.xlabel('t(s)')
plt.legend()  # note that we put labels in our plot statements to get this to work
plt.show()
```
