# Introduction
At this point, you can use Python to fit curves and lines that represent functional dependence on a single variable.<br>
Now, we take one more step: we consider functional dependence on multiple variables.  In this example, and in our DSC experiment, we will use *multiple linear regression*.  This means that we have more than one indendent variable, each that comes with its own slope term (those of you who are fans of partial derivatives can see the connection between those and these slopes).<br>

# Importing Packages & Data
Many of these steps are identical to our non-linear fitting tutorial, so the redundancy is intended.<br>
Pasting data into your scipt is not usually a great approach, and in Python this can sometimes cause more problems than it solves.<br>
When you are working with tabular data, the *pandas* package is a useful one.  That said, we will need to be aware of the file types when we use our data.  If you have not run into an error based on this before, then it is probably just a matter of time (or you are quite the python-er!).


```python
import numpy as np
import pandas as pd
from scipy import optimize
```

# File type
The most likely file formats that you will encounter here are Excel spreadsheets and comma separated value (csv) files.<br>
The former allows for pre-python calculation and column headers while the latter has less non-numeric detail but can be opened by many different applications.<br>

```python
# If you choose to use an Excel file
data_excel = pd.read_csv('files/Alkane Boiling Points.csv')
# Since you are using a spreadsheet, you can use headers.  However, our headers
# are quite long, so we will use the locations.

# Our boiling points are in the 4nd column
y_data = data_excel.iloc[:,3]

# Our independent values (# of carbons, # of branches and in the 2nd & 3rd)
x_data_1 = data_excel.iloc[:,1]
x_data_2 = data_excel.iloc[:,2]
```

# What is your function?
As before, Python will expect that your independent variable(s), here the x, will be the first entry in your function.  Any fitted parameters follow.<br>
Now, with multiple independent variables, we still pass them into the function in a single term.  The parameters follow as before.<br>
We will assume the function has a non-zero intercept and has linear dependence on each variable.

```python
# Define your fitting function
def fit_fn(x_var, a, b, c):
    nC, nB = x_var # nC = # of carbons, nB = # of branches
    y = a*nC + b*nB + c
    return y

guess = [1,1,1]

# Let's make your x-data an array
x_data = np.array([x_data_1,x_data_2])

fit_params, covar_params = optimize.curve_fit(fit_fn, x_data, y_data, guess)
```

# Interpretation
There is a large amount of information here! The parameters that have been fit appear on the left-hand side in whatever you name the first variable (here, it was *fit_params*).<br>
Recall that Python's array numbering begins with 0, so the first parameter we are fitting in the function (it appears first in the *def* statement) is at position 0.

```python
param_a = fit_params[0]
param_b = fit_params[1]
param_c = fit_params[2]

print("The a parameter is", param_a)
print("The b parameter is", param_b)
print("The c parameter is", param_c)
```

# Uncertainty
At this time, you are likely anticipating the uncertainty measurements and the
 good news is that the structure is the same: the second term on the left-hand side (here, it was *covar_params*) contains the covariance matrix.  Since we have three parameters, the uncertainties that we want are the 0,0, 1,1 and 2,2 positions.  Moreover, those diagonal covariance terms are our uncertainties, squared.  As such, we can extract them as shown below.

```python
unc_a = np.sqrt(covar_params[0,0])
unc_b = np.sqrt(covar_params[1,1])
unc_c = np.sqrt(covar_params[2,2])

print("The a parameter's uncertainty is", unc_a)
print("The b parameter's uncertainty is", unc_b)
print("The c parameter's uncertainty is", unc_c)
```

# Trust but verify
As before, it is good practice to view the data and fit together to confirm that things have worked as intended.<br>
**Warning!** The plot the you make needs to be different when you have multiple independent variables than you have in the past.  **Do not** plot y versus x (why?).  Instead, the way to observe your fitting is to plot the predicted y-values versus the actual y-values.  If the fitting is perfect, they will fall on the diagonal.  If your fitting is reasonable, you would hope to observe minor deviations both above and below the diagonal.<br>


```python
import matplotlib.pyplot as plt

# In other plots, we might make a smooth x-range to get a smooth curve.
# Here, we want a scatterplot of discrete points where each data point
# has a corresponding predicted partner.

y_predict = fit_fn(x_data, fit_params[0], fit_params[1], fit_params[2])

plt.plot(y_data, y_predict,'x')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()
```

Be aware, though, that Python (and other programs) may make your diagonal look better than it is since they will often scale the axes.  The true diagonal corresponds to axes that have the same range.<br>
You may find it helpful, too, to include that diagonal as a guide to the eye (note that this is not a fit).

```python
x_rng = np.linspace(220,400,10)

plt.plot(x_rng, x_rng, '-.', label="Perfect prediction guide to the eye")
plt.plot(y_data, y_predict,'x', label="Model predictions for training set")
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.xlim(220,400)
plt.ylim(220,400)
plt.legend()
plt.show()
```
