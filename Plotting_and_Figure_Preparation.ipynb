# **Plotting Tutorial Using matplotlib**



```python
# Import toolboxes to Python

import matplotlib.pyplot as plt
import numpy as np
```

## Generate sample sinusoidal and linear "data" and "fits" to use for this tutoriial

```python
# Generate data for this plotting tutorial


# Sinusoidal "data"
x1 = np.linspace(0, 2*np.pi, 50)         # Generate 50 evenly spaced numbers between 0 and 2pi
noise_sin = np.random.rand(50) * 0.2     # Generate random noise for each of the 50 numbers on your data set
y1 = np.sin(x1) + noise_sin              # y=sin(x) + noise     "this represents our noisy data"

# Sinusoidal "fit"             This is  not an actual fit. This can formally be done using numpy, here this is just to get a quick data set for this plot tutorial
x2 = np.linspace(0, 2*np.pi, 50)         # Generate 50 evenly spaced numbers between 0 and 2pi
y2 = np.sin(x1)                          # y=sin(x)             "this represents our perfect fit"



# Linear "data"
x3 = np.linspace(30, 90, 25)           # Generate 25 evenly spaced numbers between 30 and 90
noise_lin = np.random.rand(25) * 10    # Generate random noise for each of the 25 numbers on your data set
y3 = 0.75*x3 + 3 - noise_lin           # y=mx + b + noise       "this represents our noisy data"

# Linear "fit"                  This fit is performed using numpy.
x4 = np.linspace(30, 90, 25)           # Generate 25 evenly spaced numbers between 30 and 90

ft=np.polyfit(x3, y3, 1)               # perform 1st order linear fit to the x3 y3 "data"
slp=ft[0]                              # extract slope value from fit
int=ft[1]                              # extract intercept value from fit
y4=slp*x4+int        # y=mx + b       "this represents our perfect fit"

```

# Aside

You do not have to to run the following code block to use the reminder of this tutorial.
If the synthetic (not actual) sinusoidal "fit" in the previous code block is appalling (as it should be, but this is merely a figure generating tutorial and we just wanted to get something to plot)  you can run this code block to perform a legitimate fit on the noisy sinusoidal data.   The fit will be stored as variables x5 and y5.  If you would like to use this in the remainder of this tutorial you will have to change the (x2, y2) to (x5, y5) wherever it appears below. Alternatively, you can use this an an additional data set for practicing your plotting.

```python
# Actual nonlinear sinusoidal fit procedure

# The x1 and y1 used in this cell are from the code above, for this cell to run correctly you need to first run the
# previous "Generate data for this plotting tutorial" code block.

# This protocol or code structure can be used to performed nonlinear fits to any function you like.
# Just pay attention to what goes where when modifying to fit your nonlinear functions.

import numpy as np
from scipy.optimize import curve_fit     # We need scipy for the fitting functions

# Function to model and fit: Put your function here.  Any function of any form (not just trig. fxn's) can be modeled using this code structure
def my_function(x, A, B, C, D):       # We list the independent variable first, followed by the parameters you intend to fit
    return A * np.sin(B*x + C) + D    # Note that sin and exp don't exist in Python; we have to get them from numpy

# Fit the model to the data
parameters, covariance = curve_fit(my_function, x1, y1)     # The  x1, and y1 here are from the code above, for this cell to run correctly
                                                            # you need to first run the  "Generate data for this plotting tutorial" code block

# Parameters contains the best-fit parameters for A, B, C and D
A, B, C, D = parameters

# Covariance is the covariance matrix; the square root of its diagonal elements gives the uncertainties
A_error, B_error, C_error, D_error  = np.sqrt(np.diag(covariance))

# Display the results
print("Fitted parameters: A =", A, "   B =", B, "   C =", C, "   D =", D)
print("Uncertainties: error in A =", A_error, "   error in B =", B_error, "   error in C =", C_error, "   error in D =", D_error)


y5=my_function(x1,A,B,C,D)  # Use the determined fitting parameters, your function, and your x-vector to compute the theoretical fit vector
x5=x1

# Uncomment (remove the #) the following two lines to check the fit relative to the data
# plt.plot(x1, y1, '^', color='red' )       # Plot first data set, this is our "noisy sinusoidal data"
# plt.plot(x5, y5, '-', color='blue' )      # Plot our legitimate fit to the "noisy sinusoidal data"
```

# **Extending Displayed Data Range: Forcasting and Backcasting Data (Fits)**

```python
# These are the original data and fit with the same range in x

plt.plot(x3, y3,'s', markersize=4)            # plot "data"
plt.plot(x4, y4,'+', markersize=8)            # plot "fit"

plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
# Extend displayed fit beyond the data set in the forwards and backwards direction

# Originally the x4 was created to be the same as the x3
# x3 = np.linspace(30, 90, 25)
# x4 = np.linspace(30, 90, 25)

# To extend the fit beyond the data range, simply create a new x4 that does this: original range 30 to 90, new range 0 to 120.
x4_ext = np.linspace(0, 120, 25)

ft=np.polyfit(x3, y3, 1)         # Re-evaluate the 1st order linear fit to the x3 y3 "data" with your new extended range x4 vector
slp=ft[0]
int=ft[1]
y4_ext=slp*x4_ext+int            # y=mx + b


plt.plot(x3, y3,'r*')             # plot "data"
plt.plot(x4_ext, y4_ext,'m-.')    # plot extended "fit"

plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
# Alternatively you can display subsets or segments of your data or fits using different linestyles and colors

plt.plot(x3, y3,'co')   # plot "data"
                        # The number identified here is the element number in your vector, not the value of the number in the vector.
                        # We had vectors of length 25 so the 19 and 18 refer to the 18th and 19th element in the vector.
plt.plot(x4_ext[:19], y4_ext[:19],'y-')
plt.plot(x4_ext[18:], y4_ext[18:],'r--')        # plot extended "fit"

plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

# Creating Inset Plots: Two Plots In One!

Two methods for adding an inset plot to a main plot are covered next.
One uses *Figure.add_axes* and the other uses *Axes.inset_axes* .
Both provide similar results, and both methods require a trial and error appraoch to specify the position of the inset plot.

The position of the inset plot within the main plot follows the format [x, y, width, height]   with the coordinates being specified relative to the main axes. The lower left corner of the main axis is specified as (0, 0), and the top right corner is specified as (1,1). The width and height parameters are ratios of the size of inset plot to the main plot.  A 0.33 width means the inset plot will be 1/3 the width of the main plot and a height of 0.5 means that the inset plot will be half as tall as the main plot. Much of the positioning of your inset plot will be dependent upon the data appearing on the main plot.

```python
# Create sample data for plot and inset plot
import numpy as np

x1=np.array([2,4,6,8])
y1=np.array([10,12,17,28])
x2=np.array([1,3,5,7])
y2=np.array([20,17,8,3])
```

```python
##  Method 1:  Using Figure.add_axes


import matplotlib.pyplot as plt

# MAIN PLOT
# Create the main plot, the "parent axes"
fig, ax = plt.subplots(figsize=(6, 4)); #ax.set_box_aspect(0.5)  # Main figure and axes. The figsize is specified in inches and is width then height
ax.plot(x1, y1)  # example plot

plt.xlabel("Main plot X-axis label")
plt.ylabel("Main plot Y-axis label")
plt.legend(['Experiment 1'], loc='lower right', frameon=False)


# INSET PLOT
# Create the inset plot, the axes are relative to parent axes
inset_ax = fig.add_axes([.25, .45, .3, .4])   # [x, y, width, height]   Coordinates are specified relative to the parent axes.
                                             # Ex: (0, 0, 0.3, 0.25) will create an axes in the lower left-hand corner with width
                                             # that takes up 1/3 (0.3) of axes width and height that takes up half (0.5) of axes height.
                                             # Modify these parameters based on your data set. Trial and error until you like the look.
inset_ax.plot(x2, y2, color="r")
plt.xlabel("Inset X-axis label")
plt.ylabel("Inset Y-axis label")
plt.legend(['Experiment 2'], loc='lower left', frameon=False)

```

```python
#  Method 2:  Using Axes.inset_axes

import matplotlib.pyplot as plt

# MAIN PLOT
# Create the main plot, the "parent axes"
fig, ax = plt.subplots(figsize=(6, 4));   # Main figure and axes. The figsize is specified in inches and is width then height
ax.plot(x1, y1)  # Example plot

# Add axis labels and legend to main plot
ax.set_xlabel("Main plot X-axis label")
ax.set_ylabel("Main plot Y-axis label")
ax.legend(['Experiment 1'], loc='lower right', frameon=False)
# ax.set_title("Main Plot with Inset")

# INSET PLOT
# Create the inset plot, the axes are relative to parent axes
inset_ax = ax.inset_axes([.15, .4, .4, .5])   # [x, y, width, height]   Coordinates are specified relative to the parent axes.
                                             # Ex: (0, 0, 0.3, 0.25) will create an axes in the lower left-hand corner with width
                                             # that takes up 1/3 (0.3) of axes width and height that takes up half (0.5) of axes height.
# Add axis labels and legend to inset plot
inset_ax.plot(x2, y2, color="r")
inset_ax.set_xlabel("Inset x-axis")
inset_ax.set_ylabel("Inset y-axis")
inset_ax.legend(['Experiment 2'], loc='lower left', frameon=False)

plt.show()
```
