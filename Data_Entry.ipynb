**The preferred data entry method is below in "Entering Data Using PANDAS"**

```python
## Manually Entering Data: 1   (List to Array)


# The NumPy library is required for numerical and scientific operations in Python
import numpy as np    # Import the library as np so you don't have to type numpy every time you want to use or call the library.


# General format:   List_title = [enter your values here, each value is separated by a comma]


# "List": A list can store elements of different data types intermixed (integers, floats, Boolean, and strings)
My_list=[1,2,3,4,5]    # Here we are creating a list that only contains integers
print(My_list)         # If you want to see the output of your code use the print command


# "Array":  An array can store elements of the same data type
My_array=np.array(My_list)   # Convert your list to an "Array".  Here we call the array function in the numpy library.
print(My_array)


```

# **Using the *iloc* function in PANDAS to select data**


 **General:  INDEXING in Python starts counting with zero**


The format for this function is [row, column]   
\
 df.iloc[row, column]  
\
A colon : is used to specify "all" (or "up to and not including"). Therefore, if you want to select all of the rows or columns a colon is used instead of the column or row index

```python
## Extract a column of your data

x=df.iloc[:,0]      # Format is [row, column].
                    # Extract all ":" of the rows of column "0", and store as variable x
                    # Since column 0 is the time column, this is an alternative approach to select that column as opposed to specifying the coulmn title.
print(x)
```

```python
## Extract multiple columns of your data

# !! Remember INDEXING in Python starts counting with zero !!

# For our data set our columns are:  Zero = t (min)   One = A (%)    Two = B (%)    Three = C (%)

columns=df.iloc[:,1:3]   # Format is [row, column].
                         # Extract all ":" of the rows of column 1 up to 3 (here the colon is used to specify "up to and not including") and store as variable columns
                         # Therefore we are selecting columns 1 and 2, our A (%) and B (%) data.
print(columns)

```

```python
## Extract a row of your data

y=df.iloc[1,:]      # Format is [row, column].
                    # Extract from row "1" all ":" of the columns, and store as variable y

print(y)
```

```python
## Extract multiple rows of your data

# !! Remember INDEXING in Python starts counting with zero !!

rows=df.iloc[1:5,:]   # Format is [row, column].
                      # Extract rows 1 to 5 (here the colon is used to specify "up to and not including") of all ":" of the columns, and store as variable rows
print(rows)


# Alternatively, this operation can be performed slightly more succinctly.  Both methods yield same output.
rows_alt=df.iloc[1:5]
print(rows_alt)
```

```python
## Extract a subset of rows and columns of your data

subset=df.iloc[1:5,1:3]   # Format is [row, column].  Again here the colons are used to specify "up to and not including")
                          # Extract rows 1 to 5 (so row 1, 2, 3, and 4) of the columns 1 to 3 (so column 1, and 2) and store as variable subset
print(subset)
```

# Combine two vectors (x and y both of same length) into matrix

```python
## Combine two vectors (x and y both of same length) into matrix

# For this we also need the numpy library
import numpy as np

x=df.iloc[:,0]                  # This is a column vector
y=df.iloc[:,3]                  # This is a column vector

comb = np.array([x, y])         # This will create a matrix but with a "row vector orientation"
print('"row vector orientation"', comb)

comb_t = np.array([x, y]).T     # This will create a matrix but with a "column vector orientation"  Here we again implement the transpose operation
print('"column vector orientation"', comb_t)


# Abbreviated example to observe this more directly
# Row vs. Column vectors
row_vector = np.array([[1, 2, 3]])    # shape (1, 3)
print('row vector example', row_vector)
col_vector = np.array([[1, 2, 3]]).T  # shape (3, 1)
print('column vector example', col_vector)

# Combining into matrix
row_arr1 = np.array([4,5,6])     #  creates row vector  shape (1, 3)
row_arr2 = np.array([7,8,9])     #  creates row vector  shape (1, 3)
comb_matrix = np.array([row_arr1, row_arr2]).T  #  .T returns transpose or column vector instead or row vectors, leave this out it you want the maintain the row vector aspect
print('combined matrix', comb_matrix)
```

# "Fixing" full data set (that was defined as df) that isn't importing correctly

```python
# Working on full data set (that was defined as df)
# Assuming 'df' is your DataFrame and that you have titled columns
# (This is all easier if you just read excel in with numerical values only)
# Convert numeric columns to a numeric type
for col in df.columns:
    if df[col].dtype == 'object':  # Check if the column is of object type (likely string)
        try:
            df[col] = pd.to_numeric(df[col], errors='coerce')  # Convert to numeric, replace non-convertibles with NaN
        except:
            pass  # Skip columns that can't be converted
print(df)
```
